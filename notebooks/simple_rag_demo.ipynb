# Simple RAG Demo: ACS Scenario Study

This notebook runs a tiny retrieval pipeline against FAA-style sample study snippets included in the repo.

```python
from pilot_coach.application.services.retrieval_service import RetrievalService
from pilot_coach.infrastructure.vectorstores.local_store import LocalVectorStore

retrieval = RetrievalService(vector_store=LocalVectorStore())
```

```python
import asyncio

query = "cross-country flight planning weather minimums and risk"
results = asyncio.run(
    retrieval.get_context(
        query=query,
        top_k=3,
        filter={"certificate_type": "private_pilot_airplane"},
    )
)
results
```

```python
for i, doc in enumerate(results, start=1):
    meta = doc.get("metadata", {})
    print(
        f"{i}. {meta.get('source_title')} | {meta.get('section')} | ACS {meta.get('acs_code')}"
    )
    print(doc.get("page_content"))
    print("-" * 100)
```
